```python
import pandas as pd

df = pd.read_csv("ibm_hr.csv")

# Drop constant/irrelevant columns Power BI doesn't need
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18"])

# Handle nulls (dataset is usually clean, but just in case)
df = df.fillna({"NumCompaniesWorked": 0, "TotalWorkingYears": df["TotalWorkingYears"].median()})

# Keep categoricals as readable strings — don't encode here,
# let Power BI/DAX handle it so slicers/axes stay human-readable
df["Attrition"] = df["Attrition"].astype(str)

df.to_csv("HR_Attrition_Cleaned.csv", index=False)
```
